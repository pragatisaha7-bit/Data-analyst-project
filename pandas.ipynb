```python
import pandas as pd
import numpy as np
import re

df = pd.read_csv("anime.csv")
print("File loaded successfully!")
df.head()

def extract_episode_count(value):
    match = re.search(r'(\d+)', str(value))
    if match:
        return int(match.group(1))
    return 0

df['Episode_count'] = df['Total Episodes'].apply(extract_episode_count)


def calculate_total_months(time_str):
    years = 0
    months = 0
    y_match = re.search(r'(\d+)\s*year', str(time_str))
    m_match = re.search(r'(\d+)\s*month', str(time_str))
    if y_match:
        years = int(y_match.group(1))
    if m_match:
        months = int(m_match.group(1))
    return years * 12 + months

df['Total_Months'] = df['Total Time'].apply(calculate_total_months)

print("\nFeature extraction ho gaya!")
print(df.head())

print("\n3. Highest Score Anime:")
print(df.loc[df['Score'].idxmax()])


print("\n4. Top 5 Highest Scoring:")
print(df.sort_values(by='Score', ascending=False)[['Title', 'Score']].head(5))


print("\n5. Highest Episode Count Anime:")
print(df.loc[df['Episode_count'].idxmax()][['Title', 'Episode_count']])


print("\n6. Top 5 Episode Count:")
print(df.sort_values(by='Episode_count', ascending=False)[['Title', 'Episode_count']].head(5))


print("\n7. Longest Running Anime:")
print(df.loc[df['Total_Months'].idxmax()][['Title', 'Total Time', 'Total_Months']])
```

```python
df.to_csv("anime_final.csv", index=False)
print()
```
